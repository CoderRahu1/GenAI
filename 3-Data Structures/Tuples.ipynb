## Tuples :

## Tuples are ordered collection of items that are immutable. they are similar to list but their immutability makes them different.



````python
## Tuples

## Creating Tuples
#Tuples are created by placing a sequence of values separated by commas within parentheses. For example:
#```python

empty_tuple = ()
print(empty_tuple)  # Output: ()
single_element_tuple = (42,)  # Note the comma
print(single_element_tuple)  # Output: (42,)
tuple_with_multiple_elements = (1, 2, 3, "hello", 4.5)
# ```
# Tuples can also be created without parentheses, just by separating values with commas:
# ```python
tuple_without_parentheses = 1, 2, 3
# ```
# This is known as tuple packing, where multiple values are packed into a single tuple.
````

````python
lst = list()
print(type(lst))  # Output: <class 'list'>

tpl = tuple()
print(type(tpl))  # Output: <class 'tuple'>
````

````python
numbers = tuple([1, 2, 3, 4, 5])
print(numbers)  # Output: (1, 2, 3, 4, 5)
````

````python
numbers = tuple([1, 2, 3, 4, 5])
print(numbers)  # Output: (1, 2, 3, 4, 5)
print(type(numbers))  # Output: <class 'tuple'>
````

````python
list([1, 2, 3, 4, 5])  # This creates a list from the tuple
# Output: [1, 2, 3, 4, 5]
````

````python
mixed_tuple = (1, "hello", 3.14, True)
print(mixed_tuple)  # Output: (1, "hello", 3.14, True)
# Tuples can contain elements of different data types, making them versatile for various applications.
````

````python
## Accessing Tuple Elements
# You can access elements in a tuple using indexing, similar to lists. Indexing starts at 0:
# ```python
tuple_example = (10, 20, 30, 40, 50)
print(tuple_example[0])  # Output: 10
print(tuple_example[2])  # Output: 30
print(tuple_example[-1])        # Output: 50 (last element)
print(tuple_example[-2])       # Output: 40 (second last element)
# ```
````

````python
## Tuples Operations
# Tuples support various operations, including concatenation and repetition:
# ```python

tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
concatenated_tuple = tuple1 + tuple2
print(concatenated_tuple)  # Output: (1, 2, 3, 4, 5, 6) 
````

````python
mixed_tuple * 3
````

````python
numbers * 3
````

````python
## Immutable Nature of Tuples
# Tuples are immutable, meaning their elements cannot be changed after creation. This immutability makes tuples suitable for use as keys in dictionaries and elements in sets.
# ```python
# Attempting to modify a tuple will raise an error:
# tuple_example[0] = 100  # Raises TypeError: 'tuple' object does not support item assignment
# ```

lst = [1, 2, 3]
print(lst)  # Output: [1, 2, 3]

lst[0] = 100
print(lst)  # Output: [100, 2, 3]
````

````python
## Tuple methods 

print(numbers.count(2))  # Output: 1 (counts occurrences of 2 in the tuple)
print(numbers.index(3))  # Output: 2 (returns the index of the first occurrence of 3 in the tuple)
````

````python
### Packing and unpacking tuples
````

````python
packed_tuple = (1, 2, 3)
print(packed_tuple)  # Output: (1, 2, 3)
````

````python
## Unpacking tuples
a, b, c = packed_tuple
print(a)
````

````python
## unpacking with star operator
a, *b = packed_tuple
print(a)
````

````python
### Nested tuples
nested_tuple = (1, (2, 3), (4, 5))
print(nested_tuple)  # Output: (1, (2, 3), (4, 5))  
````

````python
## Iterating through tuples
for element in nested_tuple:
    print(element)
````

````python
## iterating through nested tuples
for outer_element in nested_tuple:
    if isinstance(outer_element, tuple):
        for inner_element in outer_element:
            print(inner_element)
    else:
        print(outer_element)
````
